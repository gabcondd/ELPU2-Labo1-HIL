## 1. Objectifs du laboratoire

Le laboratoire introduit la méthodologie **Hardware-in-the-Loop (HIL)** avec PLECS et une RT Box. La carte de contrôle exécute le code de contrôle tandis que la RT Box émule le circuit de puissance en temps réel et renvoie les mesures au contrôleur.

## 3. Blocs importants à connaître

### Blocs côté microcontrôleur

| Bloc | Rôle à retenir |
|---|---|
| ADC | Échantillonne et convertit des signaux analogiques en valeurs numériques |
| PWM | Génère les signaux PWM pour commander les interrupteurs |
| Digital In | Lit un état logique 0/1 |
| Digital Out | Génère un état logique 0/1 |
| SPI | Communication bidirectionnelle entre contrôleur et RT Box |

### Blocs côté RT Box

| Bloc | Rôle à retenir |
|---|---|
| PWM Capture | Mesure / capture un signal PWM externe |
| Digital In | Lit une entrée numérique |
| Digital Out | Génère une sortie numérique |
| Analog Out | Convertit une grandeur numérique simulée en tension analogique via DAC |
| SPI | Interface SPI côté RT Box |

## 4. Création d'un modèle HIL

1. un pour l'algorithme de contrôle ;
2. un pour le modèle physique émulé dans la RT Box.

Une adaptation de niveau est également prise en compte entre les sorties analogiques RT Box (0–5 V) et les entrées MCU (0–3,3 V), avec un gain de **0,65** dans le modèle de simulation.
